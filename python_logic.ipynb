```python
class Clock():
    def __init__(self):

        self.pins = {
            1: {"state": 0},
            2: {"state": 0},
            3: {"state": 0},
            4: {"state": 0}
        }

        self.full_set = {1, 2 , 3 , 4}

        self.clocks = {
            (0, 0, 0): {"type": "Corner",
                        "time": 12,
                        "quadrant": {1}
                        },
            (0, 1, 0): {"type": "Edge",
                        "time": 12,
                        "quadrant": {1, 2}
                        },
            (0, 2, 0): {"type": "Corner",
                        "time": 12,
                        "quadrant": {2}
                        },
            (1, 0, 0): {"type": "Edge",
                        "time": 12,
                        "quadrant": {1, 3}
                        },
            (1, 1, 0): {"type": "Centre",
                        "time": 12,
                        "quadrant": {1, 2, 3, 4}
                        },
            (1, 2, 0): {"type": "Edge",
                        "time": 12,
                        "quadrant": {2, 4}
                        },
            (2, 0, 0): {"type": "Corner",
                        "time": 12,
                        "quadrant": {3}
                        },
            (2, 1, 0): {"type": "Edge",
                        "time": 12,
                        "quadrant": {3, 4}
                        },
            (2, 2, 0): {"type": "Corner",
                        "time": 12,
                        "quadrant": {4}
                        },
            (0, 0, 1): {"type": "Corner",
                        "time": 12,
                        "quadrant": {2}
                        },
            (0, 1, 1): {"type": "Edge",
                        "time": 12,
                        "quadrant": {1, 2}
                        },
            (0, 2, 1): {"type": "Corner",
                        "time": 12,
                        "quadrant": {1}
                        },
            (1, 0, 1): {"type": "Edge",
                        "time": 12,
                        "quadrant": {2,4}
                        },
            (1, 1, 1): {"type": "Centre",
                        "time": 12,
                        "quadrant": {1, 2, 3, 4}
                        },
            (1, 2, 1): {"type": "Edge",
                        "time": 12,
                        "quadrant": {1, 3}
                        },
            (2, 0, 1): {"type": "Corner",
                        "time": 12,
                        "quadrant": {4}
                        },
            (2, 1, 1): {"type": "Edge",
                        "time": 12,
                        "quadrant": {3, 4}
                        },
            (2, 2, 1): {"type": "Corner",
                        "time": 12,
                        "quadrant": {3}
                        }
        }
    
    def change_state(self, pin_number, state):
        self.pins[pin_number]['state'] = state

    @property
    def quad_pins(self):
        quad_pins = set()
        for i in range(1, 5):
            if self.pins[i]['state'] == 1:
                quad_pins.add(i)
        return quad_pins

    @property
    def flip_quad_pins(self):
        return self.full_set - self.quad_pins

    @staticmethod
    def change_time(initial_time, move):
        return ((initial_time - 1 + move) % 12) + 1

    def update_clock(self, wheel, move, main_side):
        for (row, col, face), clock in self.clocks.items():
            if face == main_side:
                if wheel in self.quad_pins:
                    if clock['quadrant'] & (self.quad_pins):
                        clock['time'] = self.change_time(clock['time'], move)
                else:
                    if clock['type'] == 'Corner' and not (clock['quadrant'] & self.quad_pins):
                        clock['time'] = self.change_time(clock['time'], move)

        for (row, col, face), clock in self.clocks.items():
            if face != main_side:
                if wheel in self.flip_quad_pins:
                    if clock['quadrant'] & (self.flip_quad_pins):
                        clock['time'] = self.change_time(clock['time'], -move)
                else:
                    if clock['type'] == 'Corner' and not (clock['quadrant'] & self.flip_quad_pins):
                        clock['time'] = self.change_time(clock['time'], -move)
    
    def turn_wheel(self, wheel, move, main_side):            
        self.update_clock(wheel, move, main_side)

    def __str__(self):
        lines = []

        for face in [0, 1]:
            title = "LIGHT SIDE" if face == 0 else "DARK SIDE"
            lines.append(title)
            lines.append("-" * len(title))

            for r in range(3):
                row_vals = []
                for c in range(3):
                    t = self.clocks[(r, c, face)]['time']
                    row_vals.append(f"{t:2}")
                lines.append(" ".join(row_vals))

            lines.append("")

        return "\n".join(lines)
```
